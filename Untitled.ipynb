```python
def draw_board(n, m):
    for i in range(n):
        print("--- " * m)
        print("|   " * m + "|")
    print("--- " * m)
draw_board(3, 3)
```

```python
def draw_matrix(board):
    n = len(board)
    m = len(board[0])
    for i in range(n):
        print("--- " * m)
        row_display = ""
        for j in range(m):
            if board[i][j] == 1:
                row_display += "| X "
            elif board[i][j] == 2:
                row_display += "| O "
            else:
                row_display += "|   "
        print(row_display + "|")
    print("--- " * m)
```
