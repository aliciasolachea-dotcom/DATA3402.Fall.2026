```python
def draw_board(board):
    n = len(board)
    m = len(board[0])
    for i in range(n):
        print("--- " * m)
        row_display = ""
        for j in range(m):
            if board[i][j] == 1:
                row_display += "| X "
            elif board[i][j] == 2:
                row_display += "| O "
            else:
                row_display += "|   "
        print(row_display + "|")
    print("--- " * m)
#testing
sample = [[1, 0, 2],
          [0, 1, 0],
          [2, 0, 1]]
draw_board(sample)
```

```python
def check_game_state(board):
    n = len(board)

    for i in range(n):
        if all(board[i][j] == 1 for j in range(n)): return 1
        if all(board[i][j] == 2 for j in range(n)): return 2
        if all(board[j][i] == 1 for j in range(n)): return 1
        if all(board[j][i] == 2 for j in range(n)): return 2
            
    if all(board[i][i] == 1 for i in range(n)): return 1
    if all(board[i][i] == 2 for i in range(n)): return 2
    if all(board[i][n-1-i] == 1 for i in range(n)): return 1
    if all(board[i][n-1-i] == 2 for i in range(n)): return 2

    if any(0 in row for row in board): return -1
    return 0
```

```python
def make_move(board, player, row, col):
    if board[row][col] == 0:
        board[row][col] = player
        return True
    return False
#testing
b = create_board(3)
#true or false
print(make_move(b, 1, 0, 0)) 
print(make_move(b, 2, 0, 0)) 
```

```python
def draw_board_with_labels(board):
    n = len(board)
    m = len(board[0])
    cols = "   " + "  ".join(str(i+1) for i in range(m))
    print(cols)
    for i in range(n):
        print("  " + "--- " * m)
        row_display = chr(65+i) + " "
        for j in range(m):
            if board[i][j] == 1:
                row_display += "| X "
            elif board[i][j] == 2:
                row_display += "| O "
            else:
                row_display += "|   "
        print(row_display + "|")
    print("  " + "--- " * m)
```

```python
def update_board(board, player, location):
    row = ord(location[0].upper()) - 65
    col = int(location[1]) - 1
    return make_move(board, player, row, col)
```

```python
def player_turn(board, player):
    while True:
        loc = input(f"Player {player} ({'X' if player == 1 else 'O'}), enter location (e.g., A1): ")
        if update_board(board, player, loc):
            break
        print("Invalid move, try again.")
```

```python
def play_game(n=3):
    board = create_board(n)
    current_player = 1

    while True:
        draw_board_with_labels(board)
        player_turn(board, current_player)
        state = check_game_state(board)

        if state == 1:
            draw_board_with_labels(board)
            print("Player 1 (X) wins!")
            break
        elif state == 2:
            draw_board_with_labels(board)
            print("Player 2 (O) wins!")
            break
        elif state == 0:
            draw_board_with_labels(board)
            print("It’s a draw!")
            break

        current_player = 2 if current_player == 1 else 1
```

```python
play_game(5)
```

```python
import random

def computer_move(board):
    # Try to block or win
    n = len(board)
    for r in range(n):
        for c in range(n):
            if board[r][c] == 0:
                board[r][c] = 2
                if check_game_state(board) == 2:
                    return r, c
                board[r][c] = 0
    empty = [(r, c) for r in range(n) for c in range(n) if board[r][c] == 0]
    return random.choice(empty)
```

```python
def minimax(board, depth, is_maximizing):
    state = check_game_state(board)
    if state == 1: return -10 + depth
    if state == 2: return 10 - depth
    if state == 0 or depth == 0: return 0

    n = len(board)
    if is_maximizing:
        best = -float("inf")
        for r in range(n):
            for c in range(n):
                if board[r][c] == 0:
                    board[r][c] = 2
                    best = max(best, minimax(board, depth-1, False))
                    board[r][c] = 0
        return best
    else:
        best = float("inf")
        for r in range(n):
            for c in range(n):
                if board[r][c] == 0:
                    board[r][c] = 1
                    best = min(best, minimax(board, depth-1, True))
                    board[r][c] = 0
        return best
```
